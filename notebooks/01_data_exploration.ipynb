```python
import pandas as pd
import numpy as np

# Path to your data folder
path = "D:/CAPSTONE PROJECT/DATA CSV FILES/"

# Load all files
customers = pd.read_csv(path + "olist_customers_dataset.csv")
geolocation = pd.read_csv(path + "olist_geolocation_dataset.csv")
order_items = pd.read_csv(path + "olist_order_items_dataset.csv")
payments = pd.read_csv(path + "olist_order_payments_dataset.csv")
reviews = pd.read_csv(path + "olist_order_reviews_dataset.csv")
orders = pd.read_csv(path + "olist_orders_dataset.csv")
products = pd.read_csv(path + "olist_products_dataset.csv")
sellers = pd.read_csv(path + "olist_sellers_dataset.csv")
category_translation = pd.read_csv(path + "product_category_name_translation.csv")

# Put them in a dictionary so you can loop through easily
dfs = {
    'customers': customers,
    'geolocation': geolocation,
    'order_items': order_items,
    'payments': payments,
    'reviews': reviews,
    'orders': orders,
    'products': products,
    'sellers': sellers,
    'category_translation': category_translation
}

# First look at every table
for name, df in dfs.items():
    print(f"\n{'='*50}")
    print(f"TABLE: {name}")
    print(f"{'='*50}")
    print(f"Shape: {df.shape}")
    print(f"\nColumns:\n{df.columns.tolist()}")
    print(f"\nFirst 3 rows:")
    print(df.head(3))
    print(f"\nMissing values:\n{df.isnull().sum()}")
```

```python
# Step 1: Start with orders, and attach customer info
df = orders.merge(customers, on='customer_id', how='left')

# Step 2: Attach order_items (this tells us what was bought, price, etc.)
df = df.merge(order_items, on='order_id', how='left')

# Step 3: Attach product info
df = df.merge(products, on='product_id', how='left')

# Step 4: Attach category name in English (much easier to read than Portuguese)
df = df.merge(category_translation, on='product_category_name', how='left')

# Step 5: Attach seller info
df = df.merge(sellers, on='seller_id', how='left')

# Step 6: Attach payment info
df = df.merge(payments, on='order_id', how='left')

# Step 7: Attach review info
df = df.merge(reviews, on='order_id', how='left')

# Check the result
print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
print(f"Total rows: {df.shape[0]}")
print(f"Total columns: {df.shape[1]}")
print(f"\nUnique orders: {df['order_id'].nunique()}")
```

```python
# List of all date columns in your merged df
date_cols = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date',
    'shipping_limit_date',
    'review_creation_date',
    'review_answer_timestamp'
]

# Convert each one to a real datetime
for col in date_cols:
    df[col] = pd.to_datetime(df[col])

# Confirm it worked
df[date_cols].dtypes
```

```python
# Calculate delivery delay: actual delivery date minus estimated delivery date
df['delivery_delay_days'] = (df['order_delivered_customer_date'] - df['order_estimated_delivery_date']).dt.days

# Quick look at the result
df['delivery_delay_days'].describe()
```

```python
# Categorize each order as Late, Early, or On Time
def delivery_category(days):
    if pd.isna(days):
        return 'Unknown'
    elif days > 0:
        return 'Late'
    elif days == 0:
        return 'On Time'
    else:
        return 'Early'

df['delivery_category'] = df['delivery_delay_days'].apply(delivery_category)

# Count how many orders fall into each category
df['delivery_category'].value_counts()
```

```python
# Average review score for each delivery category
df.groupby('delivery_category')['review_score'].mean()
```

```python
# For each customer state, calculate % of orders that were Late
state_delay = df.groupby('customer_state')['delivery_category'].apply(
    lambda x: (x == 'Late').mean() * 100
).sort_values(ascending=False)

state_delay
```
